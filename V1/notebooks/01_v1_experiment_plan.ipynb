# V1 实验计划（准备阶段）

本 Notebook 当前只记录实验范围、输入文件和执行顺序，不包含特征计算、训练代码或实验结果。

## 当前状态
- 原始数据：已下载
- 统一行情文件：已生成
- 特征文件：仅有字段模板
- 模型训练：尚未开始
- 结果表：仅有字段模板


## 输入与输出

输入：`../data/processed/market_data_v1.csv`

计划输出：
1. `../data/processed/features_v1.csv`：完成字段计算后的特征和标签；
2. `../results/results_v1.csv`：不同模型、特征集和标的的统一评估结果；
3. 图表和运行配置：保存到 `../results/`。


## 计划执行顺序

1. 检查原始 CSV 的字段、日期、缺失值和重复记录；
2. 按 `symbol + date` 排序并生成收益、技术和相对市场特征；
3. 生成次日收益率和次日方向标签；
4. 按时间切分训练、验证和最终测试集；
5. 运行朴素基线，再运行 ARIMA、树模型和序列模型；
6. 将所有指标写入统一结果表；
7. 检查是否存在前视偏差，并在技术文档中记录运行参数。


## 重要约束

- 不随机打乱金融时间序列；
- 最终测试集在超参数冻结前不可用于模型选择；
- 所有模型使用同一套时间切分和指标；
- 特征只能使用预测时点之前可获得的信息；
- 当前不接入 SEC 或 Agentic AI 特征。

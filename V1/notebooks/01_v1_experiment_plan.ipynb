# V1 Experiment Plan (Preparation and Validation)

This Notebook records the experiment scope, input files, and execution order. It does not contain model-training code or model results.

## Current Status
- Raw market data: downloaded
- Unified market-data file: generated
- Feature file: generated by the validation pipeline
- Model training: not started
- Model result table: schema prepared only


## Inputs and Outputs

Input: `../data/processed/market_data_v1.csv`

Generated outputs:
1. `../data/processed/features_v1.csv`: calculated features and labels;
2. `../data/processed/model_ready_v1.csv`: rows with complete model inputs and labels;
3. `../data/processed/splits_v1.csv`: fixed train, validation, and test assignments;
4. `../results/`: quality reports, feature summaries, split summaries, and figures.


## Execution Order

1. Check the raw CSV fields, dates, missing values, and duplicate records;
2. Sort by `symbol + date` and generate return, technical, and market-relative features;
3. Generate next-day return and direction labels;
4. Assign fixed time-based training, validation, and test splits;
5. Run the naive baseline and later the ARIMA, tree, and sequence models;
6. Write all metrics to one result table;
7. Check for look-ahead bias and record runtime parameters in the technical document.


## Important Constraints

- Do not randomly shuffle financial time series;
- Do not use the final test set for model selection before hyperparameters are frozen;
- All models must use the same time splits and evaluation metrics;
- Features may use only information available before the prediction time;
- SEC and Agentic AI features are not included in the current validation stage.
